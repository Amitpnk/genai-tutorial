## 1. System vs. user

The system prompt sets durable behaviour: role, constraints, tone. The user turn carries the request. Mixing them — stuffing instructions into every user message — is what makes prompts drift and caching break later.

```python
SYSTEM = "You are a terse database expert. Answer in at most three sentences."

print(ask("When should I reach for a composite index?", system=SYSTEM))
```

## 2. Show, do not only tell

Two or three examples usually beat a paragraph of description. Put them in the system prompt where they stay stable across turns.

```python
# TODO: rewrite the classifier below to use three worked examples
# instead of prose instructions, then compare the outputs.
CLASSIFIER = """Classify the support ticket as: bug, billing, or feature_request.
Respond with the label only."""

tickets = [
    "The export button spins forever on Safari",
    "I was charged twice in March",
    "Any chance of a dark mode?",
]

for t in tickets:
    print(t, "->", ask(t, system=CLASSIFIER, max_tokens=20))
```

## 3. Effort

`output_config={"effort": ...}` trades cost against depth: `low` for routine classification, `high` (the default) for most work, `max` when correctness matters more than the bill. Measure the difference rather than guessing.

```python
for effort in ["low", "high"]:
    out = ask("Why might a p99 latency spike while p50 stays flat?", effort=effort)
    print(f"--- {effort} ---\n{out}\n")
```
