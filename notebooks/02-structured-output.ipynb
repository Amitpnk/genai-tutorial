## 1. The problem with parsing prose

Asking for JSON in the prompt gets you JSON *most* of the time. In production, "most" means a parse error at 3am. Constrained output makes it a guarantee.

```python
from pydantic import BaseModel

class Ticket(BaseModel):
    category: str
    severity: int
    summary: str
    needs_human: bool

response = client.messages.parse(
    model=DEFAULT_MODEL,
    max_tokens=2000,
    messages=[{"role": "user", "content": "Ticket: checkout 500s for all EU customers since the 14:02 deploy"}],
    output_format=Ticket,
)

ticket = response.parsed_output
print(ticket)
print(type(ticket.severity))  # a real int, already validated
```
