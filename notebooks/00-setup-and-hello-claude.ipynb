## 2. Your first call

The entire API is one endpoint: `POST /v1/messages`. Everything else in this tutorial — tools, RAG, agents — is a way of shaping what you send to it.

```python
response = client.messages.create(
    model=DEFAULT_MODEL,
    max_tokens=1000,
    messages=[{"role": "user", "content": "In one sentence: what is a token?"}],
)

for block in response.content:
    if block.type == "text":
        print(block.text)

print("\n", usage_line(response))
```

A response is a **list of content blocks**, not a string. Always check `block.type` before reading `.text` — once thinking and tool use enter the picture, blindly taking `content[0].text` breaks.
