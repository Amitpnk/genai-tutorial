## 1. Do you need an agent?

Four questions before you build one:

1. **Complexity** — is the task hard to specify fully in advance?
2. **Value** — does the outcome justify the latency and cost?
3. **Viability** — is the model actually good at this task?
4. **Cost of error** — can mistakes be caught and undone?

A "no" anywhere means a fixed workflow will serve you better. Most production "agents" should have been workflows.
